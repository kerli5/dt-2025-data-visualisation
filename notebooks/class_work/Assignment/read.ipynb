```python
import urllib.request

# Read text from a URL
url = "https://www.gutenberg.org/cache/epub/11/pg11.txt"   #  (public domain example)
response = urllib.request.urlopen(url)
text = response.read().decode("utf-8")

# Show first 500 characters
print(text[:500])
```

```python
# Split into words
words = text.split()

print("Total words:", len(words))
print("First 20 words:", words[:20])

# Convert all words to lowercase
words_lower = [w.lower() for w in words]

# Count how many times a specific word appears
print("Occurrences of 'monster':", words_lower.count("monster"))

# Extract only long words (length > 8)
long_words = [w for w in words_lower if len(w) > 8]
print("Sample long words:", long_words[:20])
```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Use the text loaded earlier in Task 1
stopwords = set(STOPWORDS)
stopwords.update(["project", "gutenberg", "chapter"])   # removing irrelevant words

wc = WordCloud(
    width=800,
    height=400,
    background_color="white",
    stopwords=stopwords
).generate(text)

plt.figure(figsize=(14, 7))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()

# The word cloud highlights the most frequent words in the text. Larger words appear more often.
# Like "Alice, said, little, one ..".
# Stopwords were removed so common function words like “the”, “and”, “of”.
```
