```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Food_Nutrition_Dataset.csv")
df.head()
```

```python
plt.figure(figsize=(10,5))
df['category'].value_counts().plot(kind='bar')
plt.title("Number of Foods per Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

##The column chart shows that some food categories contain many more items than others. 
##The dataset has a few categories appearing much more frequently.
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['calories'], df['carbs'], s=df['protein']*20, alpha=0.6)
plt.title("Calories vs Carbs (Bubble size = Protein)")
plt.xlabel("Calories")
plt.ylabel("Carbs")
plt.show()

##The scatter plot shows that foods with more carbs generally have higher calories. 
## Bubble sizes show protein levels, which vary of calories and carbs.
```

```python
plt.figure(figsize=(8,8))
df['category'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Category Distribution")
plt.ylabel("")
plt.show()

##The pie chart shows the proportion of each food category. 
###The largest slice belongs to Cakes and pies, meaning this category dominates the dataset
```

```python
plt.figure(figsize=(6,6))
df.boxplot(column=['calories'])
plt.title("Calories Distribution Box Plot")
plt.ylabel("Calories")
plt.show()

##The box plot reveals the spread of calorie values. Most foods fall within a moderate calorie range,
## but a few outliers have much higher calorie levels
```
